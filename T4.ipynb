```python
import pandas as pd
from google.colab import files

# Load dataset directly from uploaded CSV
uploaded = files.upload()

import pandas as pd
df = pd.read_csv("indian_house_prices.csv")

# Scale Price to ₹ Lakhs (1 Lakh = 100,000 INR)
df["Price_Lakhs"] = df["Price"] / 100000.0

print(f"Total records: {df.shape[0]}, Total features: {df.shape[1]}")
print(df[["City", "Location", "Area", "No. of Bedrooms", "Resale", "Price_Lakhs"]].head())
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    mean_absolute_percentage_error
)

# ----------------------------------------------------
# 1. Train Baseline Models for Metrics Demonstration
# ----------------------------------------------------

# Simple Linear Regression (Single Feature: Area)
X_slr = df[["Area"]].fillna(df[["Area"]].mean())
y = df["Price_Lakhs"].fillna(df["Price_Lakhs"].mean())
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_slr, y, test_size=0.2, random_state=42)

model_slr = LinearRegression().fit(X_train_s, y_train_s)
y_pred_slr = model_slr.predict(X_test_s)

# Multiple Linear Regression (5 Features)
features_mlr = ['Area', 'No. of Bedrooms', 'Resale', 'SwimmingPool', 'Gymnasium']
X_mlr = df[features_mlr].fillna(df[features_mlr].mean())
X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(X_mlr, y, test_size=0.2, random_state=42)

model_mlr = LinearRegression().fit(X_train_m, y_train_m)
y_pred_mlr = model_mlr.predict(X_test_m)

# ----------------------------------------------------
# 2. Build the Performance Metrics Matrix
# ----------------------------------------------------
def compute_regression_metrics(y_true, y_pred, k_features):
    n = len(y_true)
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mape = mean_absolute_percentage_error(y_true, y_pred) * 100
    r2 = r2_score(y_true, y_pred)
    adj_r2 = 1 - ((1 - r2) * (n - 1) / (n - k_features - 1))

    return {
        "MAE (₹ Lakhs)": round(mae, 2),
        "MSE": round(mse, 2),
        "RMSE (₹ Lakhs)": round(rmse, 2),
        "MAPE (%)": f"{round(mape, 2)}%",
        "R² Score": round(r2, 4),
        "Adjusted R²": round(adj_r2, 4)
    }

metrics_slr = compute_regression_metrics(y_test_s, y_pred_slr, k_features=1)
metrics_mlr = compute_regression_metrics(y_test_m, y_pred_mlr, k_features=len(features_mlr))

# Display as a Comparative Metrics DataFrame
performance_matrix = pd.DataFrame([metrics_slr, metrics_mlr], index=[
    "Simple Linear Regression (Area)",
    "Multiple Linear Regression (5 Features)"
])

print("=== REGRESSION PERFORMANCE MATRIX ===")
print(performance_matrix.to_markdown())
```
